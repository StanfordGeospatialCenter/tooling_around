# Reading restricted Stanford Digital Repository cloud formats with Python

This notebook collects the fixtures from `fixtures_python.ipynb` that require SDR authorization. Each cell reads or displays the data directly from a Stacks URL, but the request will fail without an authorized SDR session.

## Shapefile polygons (restricted)
PURL: https://purl.stanford.edu/bb099zb1450
This item will require SDR authorization.

```python
# Install GeoLibre for an interactive notebook map.
%pip -q install geolibre

# Import the GeoLibre map widget.
from geolibre import Map

# Define the direct SDR FlatGeobuf URL.
data_url = "https://stacks.stanford.edu/file/bb099zb1450/D31_dep.fgb"

# Create a map.
map_view = Map()

# Add the restricted cloud-optimized vector layer when authorized.
layer = map_view.add_flatgeobuf(data_url, name="SDR restricted polygon data")

# Zoom the map to the added layer's extent.
map_view.zoom_to_layer(layer)

# Display the interactive map.
map_view
```
